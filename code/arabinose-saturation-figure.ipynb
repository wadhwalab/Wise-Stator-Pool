```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

threshold = 4

file = '../swimming-cells/arabinose-saturation-data.csv'
df = pd.read_csv(file)

# Keep only motile cells
df = df[df["Mean_Speed"] >= threshold]

arabinose_groups = df.groupby("Arabinose")["Mean_Speed"].apply(list) 
speed_2 = [x for x in arabinose_groups[2] if x >= threshold] 
speed_4 = [x for x in arabinose_groups[4] if x >= threshold] 
speed_6 = [x for x in arabinose_groups[6] if x >= threshold] 
speed_8 = [x for x in arabinose_groups[8] if x >= threshold] 
speed_10 = [x for x in arabinose_groups[10] if x >= threshold]
means = [np.mean(speed_2),np.mean(speed_4),np.mean(speed_6),np.mean(speed_8),np.mean(speed_10)]
# Make swarmplot with Arabinose concentration on x-axis
plt.figure(figsize=(8,6))
plt.plot(range(5),means, marker='o', color='purple', linestyle='--', linewidth=3,markersize=10,mew=3,mec='k',zorder=2)
sns.swarmplot(data=df,x="Arabinose",y="Mean_Speed",color="#a75bd4",size=2,zorder=1)


plt.ylim(0,40)
plt.xticks([0,1,2,3,4],[])
plt.yticks([0,10,20,30,40],[])
plt.xlabel('')
plt.ylabel('')
plt.savefig('../figure-output/arabinose-saturation.pdf',dpi=600)
plt.show()
```

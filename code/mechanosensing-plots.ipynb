```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import numpy as np
import matplotlib as mpl

#Definitions

def S(N,T,b,s,B,g):
    '''unused'''
    alpha = np.exp(B*(b-T)) / (g*np.exp(B*s))
    S = alpha / (N + alpha)**2
    return S

def T_max(N,b,s,B,g):
    '''unused'''
    return b - s - (1/B)*np.log(N*g)

def r(N,T,b,s,B,g,nmax=11):
    '''analytical expression for r'''
    q = np.exp(B*(b-T)) / (N*g*np.exp(B*s))
    return 1 / (1+q)

def chi(N,T,b,s,B,g,nmax=11):
    '''analytical expression for chi'''
    num = B*N*g*np.exp(B*(b-T+s))
    den = (np.exp(B*(b-T)) + N*g*np.exp(B*s))**2
    return np.abs(num / den)
```

```python
#plotting model prediction for chi
plt.rcParams['axes.linewidth'] = 2.0  # Sets thicker axis borders globally
plt.rcParams['xtick.major.width'] = 2.0  # Sets thicker x-ticks globally
plt.rcParams['ytick.major.width'] = 2.0  # Sets thicker y-ticks globally
mpl.rcParams['axes.spines.top'] = False
mpl.rcParams['axes.spines.right'] = False

plt.figure(figsize=(8,6))
N = np.arange(0,500,0.1)

plt.plot(N,chi(N,9.5,8,3,1,0.0001),c='#9ecae1',lw=5,label='')
plt.plot(N,chi(N,12.5,8,3,1,0.0001),c='#08306b',lw=5,label='')
plt.xlim(0,500)
plt.ylim(0,0.3)
plt.yticks([0,0.3],[])
plt.xticks([0,250,500],[])

#handling legend
#handles = [Line2D([0], [0], color='#08306b',lw=5),Line2D([0], [0], color='#9ecae1',lw=5)]
#plt.legend(handles=handles,bbox_to_anchor=(0.25, 0.4), fontsize=45,frameon=False, handletextpad=0.4)

plt.tight_layout()
plt.savefig('../figure-output/chi_model.pdf', dpi=600)
```

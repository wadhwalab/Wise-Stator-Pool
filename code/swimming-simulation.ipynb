Model prediction for swimming speeds as seen in fig 1

```python
import numpy as np
import math
import random
import matplotlib.pyplot as plt
import matplotlib as mpl

plt.rcParams['axes.linewidth'] = 2.0  # Sets thicker axis borders globally
plt.rcParams['xtick.major.width'] = 2.0  # Sets thicker x-ticks globally
plt.rcParams['ytick.major.width'] = 2.0  # Sets thicker y-ticks globally
mpl.rcParams['axes.spines.top'] = False
mpl.rcParams['axes.spines.right'] = False
colors = ['#9ecae1','#4292c6','#08306b'] #colors for plot: low, med, high load

###Parameters for model
gamma = .0001
es = 3 #in units of k_BT
eb = 8 #in units of k_BT
eT = 9.5 #in units of k_BT
N_max = 11
N_D = np.arange(0,400)
B = 1

N_SS = np.zeros(400) #eT = 9.5 (low load)
for i in range (0,len(N_SS)):
    ratio = np.exp(B*(eb-eT)) / (N_D[i] * gamma * np.exp(B*es))
    N_SS[i] = N_max / (1+ratio)

eT = 12.5
N_SS2 = np.zeros(400) #eT = 13 (high load)
for i in range (0,len(N_SS)):
    ratio = np.exp(B*(eb-eT)) / (N_D[i] * gamma * np.exp(B*es))
    N_SS2[i] = N_max / (1+ratio)
    
plt.figure(figsize=(8, 6))
plt.axhline(11,c='gray')
plt.plot(N_D,N_SS,color=colors[0],lw=5)
plt.plot(N_D,N_SS2,color=colors[2],lw=5)

plt.xlim(-1,401)
plt.ylim(0,12)
plt.xticks([0,100,200,300,400],[])
plt.yticks([0,2,4,6,8,10,12],[])

plt.savefig('../figure-output/swimming-simulation.pdf', dpi=600)
#will give RunetimeWarnings for a few divide by 0s
```
